```python
import torch
from pathlib import Path
from typing import Dict, Optional
from Model import MultiModalTransformer # 替换为你的模型文件路径
def load_model_for_inference(
    checkpoint_path: str,
    model_config: dict,
    device: torch.device = None
) -> MultiModalTransformer:
    """
    最终修复版：处理_orig_mod前缀 + 过滤多余key + 加载RevIN缓冲区
    """
    # 1. 自动选择设备
    if device is None:
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    # 根据数据情况调整配置
    lob_config = model_config.get('lob_encoder', {})
    trade_config = model_config.get('trade_encoder') if 'trade' in data_dict else None
    fusion_config = model_config.get('fusion', {})
    transformer_config = model_config.get('transformer', {})
    output_config = model_config.get('output_head', {})
    # 2. 初始化模型（配置必须和保存时一致！）
    model = MultiModalTransformer(
        lob_config=lob_config,
        trade_config=trade_config,
        fusion_config=fusion_config,
        transformer_config=transformer_config,
        output_config=output_config,
        use_revin=model_config.get("use_revin", True)
    )
    model = model.to(device)
    # model._init_fusion()
    # 3. 加载checkpoint
    checkpoint = torch.load(
        checkpoint_path,
        map_location=device,
        weights_only=False
    )
    
    # 4. 步骤1：处理_orig_mod前缀
    state_dict = checkpoint["model_state_dict"]
    new_state_dict = {}
    for key, value in state_dict.items():
        if key.startswith("_orig_mod."):
            new_key = key.replace("_orig_mod.", "")
        else:
            new_key = key
        new_state_dict[new_key] = value
    
    # 5. 步骤2：过滤参数key（只保留模型中存在的可训练参数）
    model_param_keys = set(model.state_dict().keys())  # 仅包含可训练参数
    param_state_dict = {}
    buffer_state_dict = {}  # 单独存储缓冲区
    
    for key, value in new_state_dict.items():
        # 区分参数和缓冲区（RevIN的mean/std是buffer，不在model_param_keys中）
        if key in model_param_keys:
            param_state_dict[key] = value
        elif "revin" in key.lower() and (key.endswith("mean") or key.endswith("std")):
            buffer_state_dict[key] = value.to(device)  # 缓冲区转到目标设备
        else:
            print(f"⚠️  过滤多余key：{key}")
    
    # 6. 步骤3：加载可训练参数（strict=True，确保参数100%匹配）
    # 此时param_state_dict只包含模型的可训练参数，无缺失/多余
    model.load_state_dict(param_state_dict, strict=True)
    print(f"✅ 可训练参数加载完成（共{len(param_state_dict)}个）")
    
    # 7. 步骤4：手动加载RevIN缓冲区（关键！保证推理归一化正确）
    if hasattr(model, "trade_revin") and buffer_state_dict:
        # 恢复mean
        if "trade_revin.mean" in buffer_state_dict:
            model.trade_revin.mean = buffer_state_dict["trade_revin.mean"]
            print(f"✅ 恢复RevIN mean: {model.trade_revin.mean.cpu().numpy()}")
        # 恢复std
        if "trade_revin.std" in buffer_state_dict:
            model.trade_revin.std = buffer_state_dict["trade_revin.std"]
            print(f"✅ 恢复RevIN std: {model.trade_revin.std.cpu().numpy()}")
    
    # 8. 切换到推理模式（禁用Dropout/BN/RevIN训练行为）
    model.eval()
    
    print(f"\n🎉 模型加载完成！无任何匹配错误，可直接推理")
    return model
```

```python
from Train.trainer import Trainer
from Train.train_utils import (
    setup_optimizer,
    setup_scheduler,
    setup_loss_functions,
    set_seed
)
# 设置优化器和调度器
optimizer = setup_optimizer(model, train_config.get('optimizer', {}))
scheduler = setup_scheduler(optimizer, train_config.get('scheduler', {}))

# 设置损失函数
loss_fn = setup_loss_functions(train_config.get('loss', {}), device=device)

# 创建训练器
trainer = Trainer(
    model=model,
    train_loader=None,
    val_loader=test_loader,
    loss_fn=loss_fn,
    optimizer=optimizer,
    scheduler=scheduler,
    config=train_config,
    device=device
)

# 开始训练
print("=" * 50)
metrics = trainer.validate()



print(f"\n{'='*70}")
print("测试集评估结果:")
print(f"{'='*70}")
print(f"  准确率 (Accuracy):  {metrics['accuracy']:.4f}")
print(f"  精确率 (Precision): {metrics['precision_macro']:.4f}")
print(f"  召回率 (Recall):    {metrics['recall_macro']:.4f}")
print(f"  F1分数 (F1-Score):  {metrics['f1_macro']:.4f}")

print(f"\n混淆矩阵:")
print(metrics['confusion_matrix'])
print(f"{'='*70}\n")
```
